```python
# factorial # applications that are used for
# factorial of 0 is 1 # assignment
# factorial of 1 is 1
# factorial of 2 is 2 * (2-1) = 2
# factorial of 3 is 3 * 2 * 1 = 6
# factorial of 4 is 4 * 3 * 2 * 1 = 24
```

```python
def fact(numbers):
    if numbers == 0:
        return 1
    else:
        return numbers * fact(numbers - 1) # 3 * fact(3-1)
                                           # 3 * fact(2)
                                           # 3 * 2 * fact(1)
                                           # 3 * 2 * 1 * fact(0)
                                           # 3 * 2 * 1 * 1
```

```python
fact(3)
```

```python
def get_even_number(numbers):
    for i in numbers:
        if i % 2 == 0:
            return i
    return None
```

```python
get_even_number(3)
```

```python
get_even_number([1, 3, 4, 2])
```

```python
result = get_even_number([1, 3, 4, 2])
print(result)
```

```python
result = get_even_number([1, 3, 5, 7])
print(result)
```

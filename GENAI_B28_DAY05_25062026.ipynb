```python
for i in range(5):# 0.1.2.3.4
    print(f"Before if statement {i}")
    if i == 3:
        print(i)
        break
    print('I am  outside if statement')
print('No more vales present inside the for loop')
```

```python
for i in range(5):# 0.1.2.3.4
    print(f"Before if statement {i}")
    if i == 3:
        print(i)
        continue
    print('I am  outside if statement')
print('No more vales present inside the for loop')
```
